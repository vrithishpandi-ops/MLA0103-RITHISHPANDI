```python

# FIND-S Algorithm with Matplotlib
# Google Colab Program

import matplotlib.pyplot as plt
import pandas as pd

# 1. Training dataset
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes']
]

attributes = [
    'Sky', 'AirTemp', 'Humidity',
    'Wind', 'Water', 'Forecast'
]

df = pd.DataFrame(data, columns=attributes + ['EnjoySport'])

print("Training Dataset:")
print(df.to_string(index=False))

# 2. FIND-S algorithm
hypothesis = None
history = []

print("\nFIND-S Algorithm Steps:\n")

for i, sample in enumerate(data, start=1):
    if sample[-1] == 'Yes':  # Consider only positive examples
        if hypothesis is None:
            hypothesis = sample[:-1].copy()
        else:
            for j in range(len(attributes)):
                if hypothesis[j] != sample[j]:
                    hypothesis[j] = '?'

        history.append(hypothesis.copy())

        print(f"After training example {i}: {hypothesis}")

    else:
        print(f"Example {i} is negative, so it is ignored.")

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)

# 3. Visualize hypothesis changes using Matplotlib
fig, ax = plt.subplots(figsize=(11, 4))

plot_data = history
ax.axis('off')

table = ax.table(
    cellText=plot_data,
    rowLabels=[f"Step {i}" for i in range(1, len(plot_data) + 1)],
    colLabels=attributes,
    cellLoc='center',
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 2)

ax.set_title(
    "FIND-S Algorithm: Hypothesis Generalization",
    fontsize=14,
    pad=20
)

plt.tight_layout()
plt.show()

# 4. Display final hypothesis
print("\nFinal Hypothesis:")
for attr, value in zip(attributes, hypothesis):
    print(f"{attr}: {value}")
```
